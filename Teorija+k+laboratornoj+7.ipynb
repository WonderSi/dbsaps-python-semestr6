### Множества

Множество - коллекция, которая обладает следующими свойствами:
- множество не хранит порядок, поэтому не индексируется;
- множество может хранить объекты разных типов данных;
- множество изменяемо.

Создавать множества можно несколькими способами.

````python
a = set()
b = set('hello')
c = {'a', 'b', 'c', 'd'}
d = {i ** 2 for i in range(10)} # генератор множеств
e = {}  # А так нельзя!


print(f'a - {a}', f'b - {b}', f'c - {c}',
      f'd - {d}', f'e - {e}', sep='\n')
````

Из допустимых операций, не изменяющих множество, выступают:
- `len(s)` - число элементов в множестве (размер множества).
- `x in s` - принадлежит ли x множеству s.
- `set.isdisjoint(other)` - истина, если set и other не имеют общих элементов.
- `set == other` - все элементы set принадлежат other, все элементы other принадлежат set.
- `set.issubset(other)` или set <= other - все элементы set принадлежат other.
- `set.issuperset(other)` или `set >= other` - аналогично.
- `set.union(other, ...)` или `set | other | ...` - объединение нескольких множеств.
- `set.intersection(other, ...)` или `set & other & ...` - пересечение.
- `set.difference(other, ...)` или `set - other` - ... - множество из всех элементов set, не принадлежащие ни одному из other.
- `set.symmetric_difference(other)` или `set ^ other` - множество из элементов, встречающихся в одном множестве, но не встречающиеся в обоих.
- `set.copy()` - копия множества.

А это те, которые изменяют множество:
- `set.update(other, ...)`; `set |= other | ...` - объединение.
- `set.intersection_update(other, ...); set &= other & ...` - пересечение.
- `set.difference_update(other, ...); set -= other | ...` - вычитание.
- `set.symmetric_difference_update(other)`; `set ^= other` - множество из элементов, встречающихся в одном множестве, но не встречающиеся в обоих.
- `set.add(elem)` - добавляет элемент в множество.
- `set.remove(elem)` - удаляет элемент из множества. KeyError, если такого элемента не существует.
- `set.discard(elem)` - удаляет элемент, если он находится в множестве.
- `set.pop()` - удаляет первый элемент из множества. Так как множества не упорядочены, нельзя точно сказать, какой элемент будет первым.
- `set.clear()` - очистка множества.

````python
a = set('HELLO')
b = set('WORLD')

print(len(a))
print('h' in a)
print(a.isdisjoint(b))
print(a == b)
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
print(a.add('r'))
print(a.discard('o'))
print(a.remove('r'))
print(a)
print(a.pop())
print(a.clear())
print(a)
````

### Кортежи

Кортежи - коллекция, которая обладает следующими свойствами:
- кортеж хранит порядок, поэтому индексируется;
- кортеж может хранить объекты разных типов данных;
- кортеж - не изменяемый объект.

И уже не отходя от заданного стандарта, рассмотрим, как создать кортеж:

````python
a = tuple()
b = tuple('hello, world!')
c = ()
d = ('a', )
e = ('a', 'b', '12', 123, 342, 56)

print(f'a - {a}', f'b - {b}', f'c - {c}',
      f'd - {d}', f'e - {e}', sep='\n')
````

Основные операции с кортежами такие-же, как и со списками (*о которых мы поговорим чуть дальше*), но допустимы лишь те, что не приводят к изменению этого самого кортежа. 

````python
a = ('a', 'b', 'c')
b = ('d', 'e', 'f')

print(len(a))
print('d' in a)
print(a.count('a'))
print(a.index('b'))
print(a + b)
print(a * 3)
````

Так же с Python связана одна важная особенность - множественное присваивание. С его помощью некоторые действия становятся гораздо лаконичнее.

````python
a = 1
b = 3
c = 5

a, b, c = 1, 3, 5
print(f'a - {a}', f'b - {b}', f'c - {c}', sep='\n')

  a, b, c = c, a, b
# a, b, c = 5, 1, 3
print(f'a - {a}', f'b - {b}', f'c - {c}', sep='\n')
````

### Списки

Списки в Python - упорядоченные изменяемые коллекции объектов произвольных типов.

Чтобы использовать списки, их нужно создать. Создать список можно несколькими способами. Например, можно обработать любой итерируемый объект (например, строку) встроенной функцией list, или с помощью литералов, или с помощью list comprehension. Всё по ситуации.

````python
a = list()
b = []
c = list('список')
d = ['s', 'p', ['isok'], 2]
e = [c * 2 for c in 'list']
f = [c + d for c in 'list' if c != 'i' for d in 'spam' if d != 'a']

print(f'a - {a}', f'b - {b}', f'c - {c}',
      f'd - {d}', f'e - {e}', f'f - {f}', sep='\n')

f = []
for c in 'list':
    for d in 'spam':
        if c != 'i' or d != 'a':
            f.append(c + d)
````

Со списками допустимы следующие операции:
    
- `x in a` - проверка, что x содержится в а
- `x not in a` - Проверка, что x не содержится в а (то же, что и not (x in a))
- `a + a2` - конкатенация списков, то есть новый список, в котором сначала идут все элементы a, а затем все элементы a2
- `a * k` - список a, повторенный k раз
- `len(a)` - длина списка
- `a.index(x)` - индекс первого вхождения x в a (вызовет ошибку, если x not in a, то есть если х отсутствует в а)
- `a.count(x)` - количество вхождений x в a
- `a.append(x)` - добавить x в конец a
- `a.extend(a2)` - добавить элементы коллекции a2 в конец a
- `del a[n]` - удалить n-й элемент списка
- `a.clear()` - удалить из a все элементы (то же, что `del a[:]`)
- `a.copy()` - поверхностная копия a (то же, что и полный срез `a[:]`)
- `a.insert(n, x)` - вставить x в a на позицию n, подвинув последующую часть дальше
- `a.pop(n)` - получить n-й элемент списка и одновременно удалить его из списка. Вызов метода без аргументов равносилен удалению последнего элемента. 
- `a.remove(x)` - удалить первое вхождение x в a, в случае x not in a — ошибка
- `a.reverse()` - изменить порядок элементов в a на обратный (перевернуть список) (то же, что и: `a =  a[::-1]`)
- `a.sort(reverse=False/True)` - отсортировать список по возрастанию, если reverse=False и по убыванию, если reverse=True
- `bool(a)` - один из способов проверить список на пустоту (возвращает True, если список непустой, и False в противном случае)
 

````python
a = [2, 3, 5]
a2 = [2, 4, 7]

print(5 in a)
print(5 not in a)
print(a + a2)
print(a * 3)
print(len(a2))
print(a2.index(7))
print(a2.count(7))
a.append(8)
print(a)
a.extend(a2)
print(a)
del a[0]
print(a)
a3 = a.copy()
print(a3)
a3.clear()
print(a3)
a.insert(0, 8)
print(a)
a.pop(1)
print(a)
a.pop()
print(a)
a.remove(8)
print(a)
a.reverse()
print(a)
a.sort(reverse = True)
print(a)
````

#### Словари

Словари (dictionaries) в Python - неупорядоченные коллекции произвольных объектов с доступом по ключу. Их иногда ещё называют ассоциативными массивами или хеш-таблицами.

Чтобы работать со словарём, его нужно создать. Создать его можно несколькими способами.

````python
a = {}
b = {123: 1, 'dictionary': 2, (123, 234): [2342, 'fgdf', {234, 345, 345}]}
c = dict(short='dict', long='dictionary')
d = dict([(1, 1), (2, 4)])
e = dict.fromkeys(['a', 'b'])
f = dict.fromkeys(['a', 'b'], 100)
g = {a: a ** 2 for a in range(7)}

print(f'a - {a}', f'b - {b}', f'c - {c}',
      f'd - {d}', f'e - {e}', f'f - {f}',
      f'g - {g}', sep='\n')

````

````python
b = {123: 888, 'dictionary': 2, (123, 234): [2342, 'fgdf', {234, 345, 345}]}

print(b.get(122, 'Такого ключа не существует!'))
````

Посмортим, какие методы можно использовать со словарями:

- `dict.clear()` - очищает словарь.
- `dict.copy()` - возвращает копию словаря.
- `classmethod dict.fromkeys(seq[, value])` - создает словарь с ключами из seq и значением value (по умолчанию None).
- `dict.get(key[, default])` - возвращает значение ключа, но если его нет, не бросает исключение, а возвращает default (по умолчанию None).
- `dict.items()` - возвращает пары (ключ, значение).
- `dict.keys()` - возвращает ключи в словаре.
- `dict.pop(key[, default])` - удаляет ключ и возвращает значение. Если ключа нет, возвращает default (по умолчанию бросает исключение).
- `dict.popitem()` - удаляет и возвращает пару (ключ, значение). Если словарь пуст, бросает исключение KeyError. Помните, что словари неупорядочены.
- `dict.setdefault(key[, default])` - возвращает значение ключа, но если его нет, не бросает исключение, а создает ключ с значением default (по умолчанию None).
- `dict.update([other]) `- обновляет словарь, добавляя пары (ключ, значение) из other. Существующие ключи перезаписываются. Возвращает None (не новый словарь!).
- `dict.values()` - возвращает значения в словаре.

````python
d = {a: a ** 2 for a in range(7)}

d.clear()
print(d)

d = {a: a ** 2 for a in range(7)}
d1 = d.copy()
print(d1)

result = d.get(153, 'Такого элемента нет') 
print(result)

print(d.items())
print(d.keys())
print(d.values())
print(d.pop(1))
print(d.popitem())

print(d.setdefault(414, 'Значение ключа 414'))

d.update([(1, 5), (2, 10)])
print(d)
````

## Создание и вызов простейшей функции
Функции представляют блок кода, который выполняет определенную задачу и который можно повторно использовать в других частях программы. Формальное определение функции:

```python
def <имя_функции> ([параметры]):
    <тело функции>
```

Определение функции начинается с выражения def, которое состоит из имени функции, набора параметров и двоеточия. Параметры в скобках необязательны. А со следующей строки идет тело функции, т.е. набор действий, который выполняется при её вызове. Всё тело функции имеет отступ от начала строки.

Для вызова функции указывается имя функции, после которого в скобках идет передача значений для всех ее параметров. Например:

````python
def my_func():
    pass
````

````python
def My_Function():
    print('Hello, world!')

My_Function()




def say_hello(name):
    print("Привет,", name)
     
say_hello('Вася')

def say_hello(name='Мир!'):
    print("Привет,", name)
     
say_hello()
say_hello('Вася')
say_hello(name='Гриша')



def my_print(stroka, start='---', end='---'):
    print(start + stroka + end)
    
my_print('Функция работает', '8', '9', '!!!')
````

#### Позиционные и именованные аргументы

Некоторые параметры функции мы можем сделать необязательными, указав для них значения по умолчанию при определении функции.
Выглядит это следующим образом:

```python
def say_hello(name='Петя'):
    print("Привет", f', {name}' if len(name) > 0 else '', sep='')
     
say_hello()
say_hello("Вася")
```

При передаче значений функция сопоставляет их с параметрами в том порядке, в котором они передаются. При вызове функции первое значение передается первому параметру, второе значение - второму параметру, и так далее по порядку. Использование именованных параметров позволяет переопределить порядок передачи:

```python
def say_hello(name='Петя', age=18):
    print("Привет", f', {name}, возраста {age} лет.' if len(name) > 0 else '', sep='')
     
say_hello("Вася", 25)
say_hello(age=35, name="Коля")
say_hello(age=10)
```

> ***Именованные параметры предполагают указание имени параметра с присвоением ему значения при вызове функции. Если в списке параметров имеются позиционные аргументы, они должны помещаться в начало этого списка!***

### Анонимные функции, инструкция lambda

Анонимные функции могут содержать лишь одно выражение, но и выполняются они быстрее. Анонимные функции создаются с помощью инструкции lambda. В lambda-функциях не используется return, они всегда возвращают значение, которое находится внутри (после двоеточия).

````python
func = lambda x, y: x + y

print(func(1, 2))
print(func('a', 'b'))

print((lambda x, y: x + y)(1, 2))
print((lambda x, y: x + y)('a', 'b'))

def operation(x, y):
    return x + y

print(operation(5, 6))
operation = lambda x, y: x * y
print(operation(5, 6))
````

#### Неочевидное применение lambda-функций

Вы уже знаете, что для коллекций существуют такие функции как `min`, `max` и `sorted`. При помощи них можно получать максимальное значение, минимальное значение и отсортированную по возрастанию коллекцию соответственно. Но у sorted есть один не обязательный аргумент key, который позволяет творить чудеса, ведь именно он и используется для того, чтобы сортировать коллекции.

По умолчанию, значение key=None, т.е. в качестве ключей выступают сами объекты коллекции. Но посмотрите на то, как меняется порядок, если изменять этот параметр:

````python
some_numbers = [x ** 2 for x in range(1, 11)]
print(some_numbers)
print(sorted(some_numbers))
print(sorted(some_numbers, key=lambda x: x))

print(sorted(some_numbers, key=lambda x: -x))
print(sorted(some_numbers, reverse=True))
print(list(reversed(sorted(some_numbers, key=lambda x: x))))

print(sorted(some_numbers, key=lambda x: -len(str(x))))
print(sorted(some_numbers, key=lambda x: [len(str(x)), -x]))
````
