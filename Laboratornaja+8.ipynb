Используя материал занятия, документацию по библиотеке pandas и набор данных <a href='https://www.kaggle.com/gregorut/videogamesales'>Video Game Sales</a>:
1. Вывести все доступные платформы, на которых выпускалась хотя-бы одна игра;
2. Добавить к копии набора данных столбец metacritic_rating и заполнить его данными из набора данных <a href='https://www.kaggle.com/skateddu/metacritic-all-time-games-stats'>Metacritic all time games stats</a>;
3. Вывести список игр, рейтинг которых равен "M" и год издания не ранее 2012 года;
4. Рассчитать и вывести описательные статистики для списка, полученного в предыдущем пункте;
5. Вывести жанры игр с количеством игр, которые принадлежат соответствующему жанру в виде "<жанр> - <количество игр>" для всех жанров, которые содержат в своем названии не менее 3 различных гласных.

```python
import pandas as pd

df = pd.read_csv('vgsales.csv')

df.head()
```

```python
platforms = df['Platform'].unique()
print("Доступные платформы:")
for platform in sorted(platforms):
    print(platform)
```

```python
meta_df = pd.read_csv('metacritic_games.csv')

df_copy = df.copy()

meta_cols = meta_df[['name', 'metascore', 'rating']].rename(columns={
    'name': 'Name',
    'metascore': 'metacritic_rating',
    'rating': 'Rating'
})
df_copy = df_copy.merge(meta_cols, on='Name', how='left')

print("Столбцы датасета с рейтингом Metacritic:")
print(df_copy.columns.tolist())
df_copy[df_copy['metacritic_rating'].notna()].head()
```

```python
filtered = df_copy[
    (df_copy['Rating'] == 'M') &
    (df_copy['Year'] >= 2012)
]

print(f"Игры с рейтингом M (не ранее 2012 года): {len(filtered)} шт.")
filtered
```

```python
print("Описательные статистики для игр с рейтингом M (не ранее 2012 года):")
filtered.describe()
```

```python
vowels = set('aeiouAEIOU')

genre_counts = df_copy.groupby('Genre').size()

print("Жанры с не менее 3 различными гласными:")
for genre, count in genre_counts.items():
    unique_vowels = set(ch.lower() for ch in genre if ch.lower() in vowels)
    if len(unique_vowels) >= 3:
        print(f"{genre} - {count}")
```
